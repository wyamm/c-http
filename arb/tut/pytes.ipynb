```python
import pytesseract
from PIL import Image
import os
```

```python
img_file = "images/Dexsport.png"
procesed_files = ["temp/Dexsport.jpg", "temp/thunderbet_test.jpg", "temp/dexsport_small.jpg"]
```

```python
config = '--psm 6 -c tessedit_char_whitelist=ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789.+-_ '
for file in procesed_files:
    img = Image.open(file)
    ocr_result = pytesseract.image_to_string(img, config=config)
    print(file)
    print(ocr_result)
```
